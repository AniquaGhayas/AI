```python
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

tasks = ["I am hungry and need food", "I have an exam and need to study",
         "I am tired and need rest", "I need to travel somewhere",
         "I have a job interview", "I need to manage my money"]

actions = ["Find food or prepare a meal", "Prepare a study plan",
           "Take some rest", "Plan a travel route",
           "Prepare for the job interview", "Prepare a budget"]

model = TfidfVectorizer()
data = model.fit_transform(tasks)

user_task = input("Enter your task: ")
scores = cosine_similarity(model.transform([user_task]), data)[0]
best = scores.argmax()

decision = actions[best] if scores[best] > 0 else "Provide more details"
print("AI Decision:", decision)

plt.figure(figsize=(8, 4))
plt.axis("off")
plt.text(0.5, 0.8, "AI TASK ASSISTANT", ha="center", fontsize=16)
plt.text(0.5, 0.6, "Task: " + user_task, ha="center",
         wrap=True, bbox=dict(boxstyle="round", facecolor="lightblue"))
plt.text(0.5, 0.3, "Decision: " + decision, ha="center",
         wrap=True, bbox=dict(boxstyle="round", facecolor="lightgreen"))
plt.show()
```

```python
import matplotlib.pyplot as plt

cities = {
    "Hyd": (3,4), "Nalgonda": (2,2), "Warangal": (3,1),
    "Tirupati": (4,3), "Vijayawada": (5,2),
    "Kadapa": (5,5), "Nellore": (2,3), "Kurnool": (4,4)
}

connections = [("Hyd","Nalgonda"), ("Hyd","Warangal"),
    ("Hyd","Kurnool"), ("Nalgonda","Vijayawada"),
    ("Kurnool","Kadapa"), ("Kadapa","Tirupati")]

path = ["Hyd", "Nalgonda", "Vijayawada"]

plt.figure(figsize=(10,6))

for a,b in connections:
    x,y = zip(cities[a], cities[b])
    plt.plot(x,y,'b-', linewidth=2)

for a,b in zip(path, path[1:]):
    x,y = zip(cities[a], cities[b])
    plt.plot(x,y,'r--', linewidth=3)

for city,(x,y) in cities.items():
    color = 'red' if city in ["Hyd","Vijayawada"] else 'yellow'
    size = 500 if city == "Hyd" else 200
    plt.scatter(x,y,s=size,c=color,edgecolor='blue',zorder=5)
    plt.text(x+0.15,y+0.1,city,fontsize=10)

plt.title("STATE-BASED DECISION MAKING AGENT")
plt.text(2.5,7,"Initial: Hyd | Goal: Vijayawada")
plt.show()
```

```python
import matplotlib.pyplot as plt

graph = {
    'A':['B','C'], 'B':['D','E'], 'C':['F'],
    'D':['G'], 'E':['G'], 'F':['G'], 'G':[]
}

h = {'A':6, 'B':4, 'C':5, 'D':2, 'E':3, 'F':4, 'G':0}

pos = {
    'A':(0,3), 'B':(-2,2), 'C':(2,2),
    'D':(-3,1), 'E':(-1,1), 'F':(2,1), 'G':(0,0)
}

start, goal = 'A', 'G'
current, path = start, [start]

while current != goal and graph[current]:
    next_node = min(graph[current], key=lambda x: h[x])
    if h[next_node] < h[current]:
        current = next_node
        path.append(current)
    else:
        break

print("Path:", " -> ".join(path))

plt.figure(figsize=(10,7))

for n in graph:
    for v in graph[n]:
        plt.annotate("", xy=pos[v], xytext=pos[n],
                     arrowprops=dict(arrowstyle="->", color="gray",
                                     shrinkA=25, shrinkB=25))

for a,b in zip(path,path[1:]):
    plt.annotate("", xy=pos[b], xytext=pos[a],
                 arrowprops=dict(arrowstyle="->", color="red",
                                 lw=3, shrinkA=25, shrinkB=25))

for n,(x,y) in pos.items():
    color = "lightgreen" if n == goal else "orange" if n in path else "lightblue"
    plt.scatter(x,y,s=1800,c=color,edgecolor="black",zorder=3)
    plt.text(x,y,f"{n}\n(h={h[n]})",ha="center",
             va="center",fontsize=12,fontweight="bold",zorder=4)

plt.title("Hill Climbing Algorithm")
plt.xlim(-4.5,3.5)
plt.ylim(-1.5,4)
plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
import heapq
import networkx as nx
import matplotlib.pyplot as plt

graph = {
    'A': [('B',2),('C',5)], 'B': [('D',4),('E',1)],
    'C': [('F',2)], 'D': [('G',3)],
    'E': [('G',6)], 'F': [('G',1)], 'G': []
}

def ucs(start, goal):
    pq = [(0, start, [start])]
    visited = set()

    while pq:
        cost, node, path = heapq.heappop(pq)
        if node in visited:
            continue
        visited.add(node)

        if node == goal:
            return path, cost

        for n, w in graph[node]:
            if n not in visited:
                heapq.heappush(pq, (cost+w, n, path+[n]))

path, cost = ucs('A', 'G')
print("UCS Path:", " -> ".join(path))
print("UCS Cost:", cost)

G = nx.Graph()
for n in graph:
    for v, w in graph[n]:
        G.add_edge(n, v, weight=w)

pos = {'A':(0,2), 'B':(-2,1), 'C':(2,1),
       'D':(-3,0), 'E':(-1,2), 'F':(2,0), 'G':(0,-1)}

nx.draw(G, pos, with_labels=True, node_size=1800)
nx.draw_networkx_edge_labels(G, pos, nx.get_edge_attributes(G,'weight'))
nx.draw_networkx_edges(G, pos, edgelist=list(zip(path,path[1:])),
                       edge_color='red', width=3)

plt.title("Uniform Cost Search")
plt.xlim(-4,3)
plt.ylim(-2,3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import networkx as nx

values = [3, 5, 6, 9, 1, 2, 0, -1]

def minimax(node, depth, alpha, beta, is_max):
    if depth == 3:
        return values[node]

    best = float('-inf') if is_max else float('inf')

    for i in range(2):
        v = minimax(node*2+i, depth+1, alpha, beta, not is_max)

        if is_max:
            best = max(best, v)
            alpha = max(alpha, best)
        else:
            best = min(best, v)
            beta = min(beta, best)

        if alpha >= beta:
            print("Branch pruned at node", node)
            break

    return best

result = minimax(0, 0, float('-inf'), float('inf'), True)
print("Optimal value:", result)

# Draw game tree
tree = {
    'A':['B','C'], 'B':['D','E'], 'C':['F','G'],
    'D':['3','5'], 'E':['6','9'],
    'F':['1','2'], 'G':['0','-1']
}

G = nx.DiGraph()
for p, children in tree.items():
    for c in children:
        G.add_edge(p, c)

pos = {
    'A':(0,3), 'B':(-2,2), 'C':(2,2),
    'D':(-3,1), 'E':(-1,1), 'F':(1,1), 'G':(3,1),
    '3':(-3.5,0), '5':(-2.5,0), '6':(-1.5,0),
    '9':(-0.5,0), '1':(0.5,0), '2':(1.5,0),
    '0':(2.5,0), '-1':(3.5,0)
}

labels = {n: n if n not in ['A','B','C','D','E','F','G']
          else n + (' MAX' if n in ['A','D','E','F','G'] else ' MIN')
          for n in G.nodes()}

nx.draw(G, pos, labels=labels, with_labels=True,
        node_size=1600, node_color='lightblue',
        font_size=10, font_weight='bold')
plt.title("Minimax with Alpha-Beta Pruning")
plt.axis('off')
plt.show()
```
